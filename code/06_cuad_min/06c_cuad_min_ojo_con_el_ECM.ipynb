```python
import numpy as np #para manejar arreglos
from matplotlib import pyplot as plt #para graficar

from numpy.polynomial import Polynomial as Poly

plt.rcParams["figure.figsize"] = (8,5) #tamaño de los gráficos: ancho x alto
```

# ¿Es el mejor modelo el de menor ECM?

```python
def f(x):
  y=1-x**2
  return y

a = 20
tabla_x = np.linspace(-1,1,a)
y_orig = f(tabla_x)
std = np.std(y_orig)*0.2
tabla_y = f(tabla_x) + np.random.normal(0,std,a)

plt.plot(tabla_x, tabla_y, '*', label='Datos')
plt.legend()
plt.grid()
```

```python
plt.plot(tabla_x, tabla_y, '*', label='Datos')

x = np.linspace(-1,1,1000)
grados = [2,10,15]

for grado in grados:
  cuad_min = Poly.fit(tabla_x, tabla_y, deg=grado)

  vals_y_cuad_min = cuad_min(tabla_x)
  cant_datos = len(tabla_x)
  ECM = (1/cant_datos)*np.linalg.norm(vals_y_cuad_min - tabla_y, ord=2)**2
  print("Error cuadrático medio del modelo",grados.index(grado),":", ECM)

  plt.plot(x, cuad_min(x), label='Modelo {}'.format(grados.index(grado)))

plt.legend()
plt.grid()
# b = 0.2
# plt.ylim(-b,1+b)
```
