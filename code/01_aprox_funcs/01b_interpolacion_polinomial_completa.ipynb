## Interpolar una función evaluada en puntos, y estimar el máximo error
- ### Interpolar $sen(\pi x)$ con 3 puntos equiespaciados entre 0 y 0.5
- ### Graficar $sen(\pi x)$ y el polinomio interpolador, para x entre 0 y 0.5
- ### Estimar el máximo error entre $sen(\pi x)$ y el polinomio interpolador, para x entre 0 y 0.5 (como el ejercicio 10)

```python
#defino la función
def funcion_sen(x):
    y = np.sin(np.pi*x)
    return y

#defino la tabla de valores para interpolar
tabla2_x = np.linspace(0, 0.5, 3) #desde 0 y hasta 0.5, con 3 valores equiespaciados
tabla2_y = funcion_sen(tabla2_x)


# A PARTIR DE ACÁ ES IGUAL QUE ANTES
grado2 = len(tabla2_x) - 1 #el grado siempre es (a lo sumo) uno menos que la cantidad de datos
# calculo el polinomio interpolador
polinomio_interpolador2 = Poly.fit(tabla2_x, tabla2_y, grado2)


# graficar función y pol interp
vals_x_graficar2 = np.linspace(0,0.5,10**5) #¡muchos! para que parezca una línea continua. Grafico en el rango de interpolación.
vals_y_polinterp_graficar = polinomio_interpolador2(vals_x_graficar2)
vals_y_func_graficar = funcion_sen(vals_x_graficar2)

#grafico tabla que usé para interpolar
plt.plot(tabla2_x, tabla2_y, '*', label = "tabla interp") # el * es para graficar solamente el punto

# grafico el polinomio interpolador
plt.plot(vals_x_graficar2, vals_y_polinterp_graficar, label = "pol interp")

# grafico la función original
plt.plot(vals_x_graficar2, vals_y_func_graficar, label = "sen(pi*x)")

plt.legend() #muestra etiquetas


# estimar error máximo entre función y pol interp
diferencias = vals_y_func_graficar - vals_y_polinterp_graficar #vector diferencia entre func y pol interp
error = np.linalg.norm(diferencias, ord=np.inf) #norma infinito: el maximo de los módulos de cada elemento
print("El error máximo (estimado) entre la func y el pol interp es:", error)
# ¿Es grande o es chico el error?
```
