### ¿Cómo encontramos el polinomio interpolador? Que nos resuelva el problema `Python`: en la 1er. celda ya importamos lo que necesitamos, **¡no olvidarse de ejecutarla!**.


```python
# creo la tabla de valores a interpolar
tabla_x = np.array( [-2,-1,2,3] )
tabla_y = np.array( [4,1,4,9] )

#calculo el grado del polinomio interpolador
# len calcula la longitud de un vector, es decir cantidad de componentes
grado = len(tabla_x) - 1

# calculo el polinomio interpolador
polinomio_interpolador = Poly.fit(tabla_x, tabla_y, grado).convert() # el .convert() es para que sea base canónica, sino por defecto una base Cheby por tema estabilidad numérica
print("El polinomio interpolador es:", polinomio_interpolador)
print("Sus coeficientes son (orden creciente de potencias de x):", polinomio_interpolador.coef)


#ahora este pol interp lo puedo evaluar en uno o varios valores, al igual que las funciones que definimos en la unidad 00
print("Polinomio interpolador evaluado en x=1.5:", polinomio_interpolador(-1.5))
print("Polinomio interpolador evaluado en los x que usamos para interpolar:", polinomio_interpolador(tabla_x))


# Cosas extra:
# al polinomio lo puedo derivar
print("\nCosas extra:")
pol_interp_derivado = polinomio_interpolador.deriv()
print("Polinomio interpolador:", polinomio_interpolador)
print("Derivada del polinomio interpolador:", pol_interp_derivado)

# le puedo estimar las raíces
print("")
raices_estimadas = polinomio_interpolador.roots()
print("Raíces estimadas:", raices_estimadas)
print("¡OJO, DAN MAL las raíces! ¿Por qué?")

print(" ")
print("En cambio si defino el ponomio a mano, dan bien las raíces")
pol_x2 = Poly([0,0,1])
print("Polinomio:", pol_x2)
print("Raíces estimadas:", pol_x2.roots())

# y se pueden hacer muchas cosas más, mirar acá
# https://numpy.org/doc/stable/reference/routines.polynomials.classes.html
```
