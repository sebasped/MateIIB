## Definir una función `paso_biseccion` que haga **UN SOLO PASO** del método de bisección

### ¿Qué tiene que recibir como entrada esta función? ¿Qué tiene que devolver?

```python
def paso_biseccion(extremo_izq, extremo_der, funcion):

    punto_medio = (extremo_izq + extremo_der)/2
    func_punto_medio = funcion(punto_medio)
    func_ext_izq = funcion(extremo_izq)

    if func_punto_medio == 0:
      extremo_izq_sig = punto_medio
      extremo_der_sig = punto_medio

    elif func_ext_izq*func_punto_medio > 0:
      extremo_izq_sig = punto_medio
      extremo_der_sig = extremo_der

    else:
      extremo_izq_sig = extremo_izq
      extremo_der_sig = punto_medio

    return extremo_izq_sig, extremo_der_sig
```

### ¿Cómo podemos elegir el intervalo inicial?

```python
def func(x):
    y = x**2-2
    return y

a = 0
b = 2
print("El intervalo inicial es el:", a, b)
print("La aproximación inicial es:", (a+b)/2)
print("La cota del error es:", abs(a-b)/2)

# primer paso
nuevo_a, nuevo_b = paso_biseccion(a, b, func)
print("")
print("El intervalo luego del primer paso es:", nuevo_a, nuevo_b)
print("La aproximación es:", (nuevo_a+nuevo_b)/2)
print("La cota del error es:", abs(nuevo_a-nuevo_b)/2)

# segundo paso
a = nuevo_a
b = nuevo_b
nuevo_a, nuevo_b = paso_biseccion(a, b, func)
print("")
print("El intervalo luego del segundo paso es:", nuevo_a, nuevo_b)
print("La aproximación es:", (nuevo_a+nuevo_b)/2)
print("La cota del error es:", abs(nuevo_a-nuevo_b)/2)

# tercer paso
a = nuevo_a
b = nuevo_b
nuevo_a, nuevo_b = paso_biseccion(a, b, func)
print("")
print("El intervalo luego del tercer paso es:", nuevo_a, nuevo_b)
print("La aproximación es:", (nuevo_a+nuevo_b)/2)
print("La cota del error es:", abs(nuevo_a-nuevo_b)/2)

# cuarto paso
a = nuevo_a
b = nuevo_b
nuevo_a, nuevo_b = paso_biseccion(a, b, func)
print("")
print("El intervalo luego del cuarto paso es:", nuevo_a, nuevo_b)
print("La aproximación es:", (nuevo_a+nuevo_b)/2)
print("La cota del error es:", abs(nuevo_a-nuevo_b)/2)

# quinto paso
a = nuevo_a
b = nuevo_b
nuevo_a, nuevo_b = paso_biseccion(a, b, func)
print("")
print("El intervalo luego del quinto paso es:", nuevo_a, nuevo_b)
print("La aproximación es:", (nuevo_a+nuevo_b)/2)
print("La cota del error es:", abs(nuevo_a-nuevo_b)/2)

print("")
print("Y así siguiendo... ¿parece que se acerca a lo que se tiene que acercar?, ¿qué error estoy cometiendo?")

print("El valor que nos da Python es:", 2**(1/2))
```

## Definir `metodo_biseccion_cant` que aplique `paso_biseccion` una cantidad de veces fijada.
