# Método de Newton-Raphson: ¿cuál es la idea y objetivo de este método?
---

## Definir `metodo_newtonRaphson_tol` que aplique el método de Newton-Raphson la cantidad de veces _necesaria_ hasta que $|x_{n+1} - x_n| < ϵ$, donde $ϵ$  es la tolerancia y es típicamente un número chico.

---

## Idea igual que antes:
### **Primero** definir `paso_newtonRaphson` que haga **UN SOLO PASO** de Newton-Raphson
### ¿Qué tiene que recibir como entrada esta función? ¿Qué tiene que devolver?


### **Segundo** definir `metodo_newtonRaphson_tol`.
### ¿Qué tiene que recibir como entrada esta función? ¿Qué tiene que devolver?


```python
def paso_newtonRaphson(x, f, df):

    x_sig = x - f(x)/df(x)

    return x_sig


def metodo_newtonRaphson_tol(x_inic, f, df, tolerancia_deseada):
    x = x_inic

    tolerancia_actual = tolerancia_deseada + 1 #para que entre al while, pues no puedo calcular la tolerancia actual solamente con x_inic
    cant_pasos = 0
    while tolerancia_actual >= tolerancia_deseada:
        x_sig = paso_newtonRaphson(x, f, df)
        tolerancia_actual = abs(x_sig - x) #recalculo la tolerancia actual
        x = x_sig
        cant_pasos = cant_pasos + 1

    print("Cantidad total de pasos:", cant_pasos)
    return x
```

### Testeamos  `metodo_newtonRaphson_tol` para aproximar una solución de $x^2-2=0$.

```python
# Testeamos Newton_Raphson con un ejemplo sencillo

def f(x):
    y = x**2-2
    return y

def df(x):
    y = 2*x
    return y

x_inic = 5 #experimentar variando
tolerancia = 10**(-5)  #experimentar variando

x_final = metodo_newtonRaphson_tol(x_inic, f, df, tolerancia)
print("La aproximación con tolerancia", tolerancia, "es:", x_final )
print("El valor que da Python es:", 2**(1/2))
print("¿Qué error estoy cometiendo? Comparando con el valor que da Python:", abs(x_final - 2**(1/2)))
print("")
print("¿Cuántos pasos en total se dieron? Lo imprime metodo_newtonRaphson_tol (primera línea).")

# bisección en [0,2]: necesito 2/2**n < tolerancia, es decir n > log2(2/tolerancia)
pasos_biseccion = int(np.ceil(np.log2((2-0)/tolerancia)))
print("¿Necesité más o menos pasos que en bisección para la misma tolerancia? Bisección en [0, 2] necesita:", pasos_biseccion, "pasos")
```

```python
# Método de la Secante: igual que Newton-Raphson pero cambiando la derivada por
# la pendiente de la recta secante que pasa por los dos últimos puntos

def paso_secante(x_ant, x, f):

    x_sig = x - f(x)*(x - x_ant)/(f(x) - f(x_ant))

    return x_sig


def metodo_secante_tol(x0, x1, f, tolerancia_deseada):
    x_ant = x0
    x = x1

    tolerancia_actual = abs(x - x_ant)
    cant_pasos = 0
    while tolerancia_actual >= tolerancia_deseada:
        x_sig = paso_secante(x_ant, x, f)
        tolerancia_actual = abs(x_sig - x) #recalculo la tolerancia actual
        x_ant = x
        x = x_sig
        cant_pasos = cant_pasos + 1

    print("Cantidad total de pasos:", cant_pasos)
    return x


# lo aplico al ejemplo de antes: x**2-2=0
x0 = 5
x1 = 4
tolerancia = 10**(-5)

x_final = metodo_secante_tol(x0, x1, f, tolerancia)
print("La aproximación por Secante con tolerancia", tolerancia, "es:", x_final)
print("El valor que da Python es:", 2**(1/2))
print("Error respecto al valor que da Python:", abs(x_final - 2**(1/2)))
```

```python
# Método de Regula Falsi: igual que bisección, pero en vez del punto medio uso
# el punto donde la recta secante entre los extremos corta al eje x

def paso_regulaFalsi(extremo_izq, extremo_der, funcion):

    func_ext_izq = funcion(extremo_izq)
    func_ext_der = funcion(extremo_der)
    punto_corte = extremo_izq - func_ext_izq*(extremo_der - extremo_izq)/(func_ext_der - func_ext_izq)
    func_punto_corte = funcion(punto_corte)

    if func_punto_corte == 0:
      extremo_izq_sig = punto_corte
      extremo_der_sig = punto_corte

    elif func_ext_izq*func_punto_corte > 0:
      extremo_izq_sig = punto_corte
      extremo_der_sig = extremo_der

    else:
      extremo_izq_sig = extremo_izq
      extremo_der_sig = punto_corte

    return extremo_izq_sig, extremo_der_sig, punto_corte


def metodo_regulaFalsi_cant(ext_izq, ext_der, func, cant_pasos):
    a = ext_izq
    b = ext_der

    for i in range(0,cant_pasos,1):
        nuevo_a, nuevo_b, aprox = paso_regulaFalsi(a, b, func)
        a = nuevo_a
        b = nuevo_b
        # print("Paso:", i, "Intervalo:", a, b, "Aproximación:", aprox) #descomentar para ver cada paso

        if a == b: #encontré la raíz exacta, no sigo (sino divido por cero)
            break

    return aprox


# lo testeo con el ejemplo de siempre: x**2-2=0
a = 0
b = 2
cantidad = 20 # experimentar con diferentes valores

x_final = metodo_regulaFalsi_cant(a, b, f, cantidad)
print("La aproximación por Regula Falsi luego de", cantidad, "pasos es:", x_final)
print("El valor que da Python es:", 2**(1/2))
print("Error respecto al valor que da Python:", abs(x_final - 2**(1/2)))
# Observar: a diferencia de bisección, el intervalo NO se achica a cero (el extremo derecho queda fijo en 2),
# por eso no conviene usar |b_n - a_n| como criterio de parada.
```
