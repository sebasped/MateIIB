# _Estrategia de codeo_

## **Primero**: definir una función que haga UN SOLO PASO del método

## **Segundo**: definir una función que utilice la anterior muchas veces, es decir haga muchos pasos y por lo tanto efectivamente aplique el método

### Comentario: la cantidad de pasos totales puede estar fijada previamente, dependiente de la tolerancia que se desee, u otras cosas.

---

```python
#z y z_sig son vectores
def paso_puntoFijo_2D(z, func):
  z_sig = func(z)
  return z_sig

#z0, z y z_sig son vectores
def metodo_puntoFijo_2D_cant(z0, func, cant_pasos):
  z = z0

  for i in range(0,cant_pasos,1):
    z_sig = paso_puntoFijo_2D(z, func)
    z = z_sig

  return z
```

# Método de Newton-Raphson para sistemas

## Consideremos el sistema de ecuaciones no lineales:
$$
 \begin{cases}
    e^{x+y} -x &= 3 \\
    x^3+2xy &= 1 \\
 \end{cases}
$$

1. ## Escribir en papel el método de Newton-Raphson: la función, la matriz diferencial, y la iteración.
1. ## Definir la función `metodo_newtonRaphson_2D_tol` que implemente el método de Newton-Raphson para sistemas, parando cuando $||v_n - v_{n-1}||$ sea menor a una tolerancia dada.
  - ### La estrategia de siempre: primero una función que realice un único paso, y luego una que dé muchos pasos.
1. ## Utilizando la función anterior buscar soluciones al sistema comenzando en $(x_0,y_0)=(-1,1)$ y en $(x_0,y_0)=(-1,-1)$. ¿Cuántos pasos fueron necesarios en cada caso?

```python
#z0, z y z_sig son vectores
def paso_newtonRaphson_2D(z, func, mat_dif):

  mat_dif_inv = np.linalg.inv(mat_dif(z))
  z_sig = z - np.matmul(mat_dif_inv, func(z) )
  # https://numpy.org/doc/stable/reference/generated/numpy.matmul.html#numpy.matmul
  # https://numpy.org/doc/stable/reference/generated/numpy.dot.html

  return z_sig


def metodo_newtonRaphson_2D_tol(z0, func, mat_dif, tol_deseada):

  z = z0
  # cant_pasos = 0 # descomentar para saber la cantidad total de pasos

  tol_actual = tol_deseada + 1 #para que entre al while, pues no puedo calcular la tolerancia actual solamente con z0
  while tol_actual >= tol_deseada:
    z_sig = paso_newtonRaphson_2D(z, func, mat_dif)
    # cant_pasos = cant_pasos + 1 # descomentar para saber la cantidad total de pasos
    tol_actual = np.linalg.norm( z_sig - z ) #recalculo la tolerancia actual
    z = z_sig

  # print('Cantidad total de pasos', cant_pasos) # descomentar para saber la cantidad total de pasos
  return z
```

```python
# test del método de newton en Rn

# defino la función: z y w son vectores
def F(z):
  x = z[0]
  y = z[1]
  w = np.array( [np.exp(x+y)-x-3, x**3+2*x*y-1] )
  return w

# defino la matriz diferencial: z es un vector
def DF(z):
  x = z[0]
  y = z[1]
  matriz = np.array( [ [np.exp(x+y)-1, np.exp(x+y)], [3*x**2+2*y, 2*x] ] )
  return matriz


z0 = np.array([-1,1])
tolerancia = 10**(-10)
z_final = metodo_newtonRaphson_2D_tol(z0, F, DF, tolerancia)
print(z_final)

print('Chequeo convergencia (tiene que dar cero)', F(z_final)) # tiene que dar cero



z0 = np.array([-1,-1])
tolerancia = 10**(-10)

z_final = metodo_newtonRaphson_2D_tol(z0, F, DF, tolerancia)
print(z_final)

print('Chequeo convergencia (tiene que dar cero)', F(z_final)) # tiene que dar cero

# # Observar que convergen a sols distintas
```
