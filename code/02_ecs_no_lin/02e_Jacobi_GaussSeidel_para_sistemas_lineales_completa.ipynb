# Jacobi y Gauss-Seidel (punto fijo para sistemas lineales)

## Pensando a Jacobi o a Gauss-Seidel como métodos de punto fijo (en varias variables) y despejando, se obtienen las fórmulas para las iteraciones de cada método.

### Está explicado muy bien en el apunte "Elementos de Cálculo Numérico" de Ricardo G. Durán, Silvia B. Lassalle y Julio D. Rossi, en el Capítulo 3 Sección 2: [https://mate.dm.uba.ar/%7Eslassall/Duran_Lassalle_Rossi.pdf](https://mate.dm.uba.ar/%7Eslassall/Duran_Lassalle_Rossi.pdf)

## Jacobi

Para resolver el sistema
$$
A{x} ={b}
$$
utilizamos la iteración
$$
x^{(k+1)}= B_J x^{(k)} + c_J $$
donde $B_J =-D^{-1} (L+U)$ y $c_J=D^{-1}b$

**Ejemplo:** Usamos el método de Jacobi (en forma matricial) para resolver el sistema $Ax = b$ con
$$
A =
\begin{pmatrix}
4 & 1 \\
2 & 3
\end{pmatrix}
\quad \mbox{y} \quad
b =\begin{pmatrix}
5 \\
5
\end{pmatrix}
$$


```python
# defino jacobi
def jacobi(A, b, x0, cant_iter, tol): #ojo x0 es un vector

  diag_vect = np.diag(A)
  D = np.diag(diag_vect)
  L = np.tril(A, -1)
  U = np.triu(A, +1)
  D_inv = np.linalg.inv(D)
  B_J = -np.dot(D_inv, (L+U) )

  c_J = np.dot(D_inv, b)
  x = x0

  error = tol + 10
  for i in range(cant_iter):
    x_sig = np.dot(B_J, x) + c_J

    error = np.linalg.norm(x_sig - x)
    if (error < tol):
      return x_sig

    x = x_sig

  return x
```

```python
# lo testeo para el ejemplo anterior

A = np.array( [[4,1], [2,3]] )
b = np.array( [5,5] )

x0 = np.array( [0,0] ) #vector inicial

x_final = jacobi(A, b, x0, 50, 10**(-5))

print('Solución Jacobi', x_final)
print('Solución de numpy:', np.linalg.solve(A,b))
```

## Gauss-Siedel
Para resolver el sistema
$$
A{x} ={b}
$$
utilizamos la iteración
$$
{x} ^{(k+1)}= B_{GS}{x}^{(k)} + c_{GS}
$$

con $B_{GS}=-(L+D)^{-1}U$ y $c_{GS}=(L+D)^{-1}b$



```python
def gauss_seidel(A, b, x0, cant_iter, tol):

  diag_vect = np.diag(A)
  D = np.diag(diag_vect)
  L= np.tril(A, -1)
  U = np.triu(A, +1)
  DL_inv = np.linalg.inv(D+L)
  B_GS = -np.dot(DL_inv, U)

  c_GS = np.dot(DL_inv, b)

  x = x0
  error = tol + 10
  for i in range(cant_iter):
    x_sig = np.dot(B_GS, x) + c_GS

    error = np.linalg.norm(x_sig - x)
    if (error < tol):
      return x_sig

    x = x_sig

  return x
```

```python
# lo testeo para el ejemplo anterior

A = np.array( [[4,1], [2,3]] )
b = np.array( [5,5] )

x0 = np.array( [0,0] ) #vector inicial

x_final = gauss_seidel(A, b, x0, 50, 10**(-5))

print('Solución Gauss-Seidel', x_final)
print('Solución de numpy:', np.linalg.solve(A,b))
```

# OJO: Jacobi y Gauss-Seidel no siempre funcionan (i.e. convergen)

## Hay casos donde ninguno funciona, hay casos donde Jacobi funciona y GS, y hay casos donde GS funciona y Jacobi no.

```python
#EJEMPLO de Jacobi no funciona y Gauss-Seidel sí

A = np.array( [[29, 2, 1], [2, 6, 1], [1, 1, 1/5]] )

b = np.array([1,2,3])
x0 = np.random.rand(3)


x_final_J = jacobi(A, b, x0, 50, 10**(-5))
print('Sol Jacobi', x_final_J)

x_final_GS = gauss_seidel(A, b, x0, 100, 10**(-2))
print('Sol Gauss-Seidel', x_final_GS)

print('Sol de numpy:', np.linalg.solve(A,b))
```

### ¿Podemos de antemano saber si van a converger?

Sí, mirando los módulos de los autovalores de la matriz del método ($B_J$ o $B_{GS}$): si son todos en módulo menores a 1, el método converge.

```python
A = np.array( [[29, 2, 1], [2, 6, 1], [1, 1, 1/5]] )

diag_vect = np.diag(A)
D = np.diag(diag_vect)
L = np.tril(A, -1)
U = np.triu(A, +1)
D_inv = np.linalg.inv(D)
B_J = -np.dot(D_inv, L+U)
eJ=np.linalg.eig(B_J)
print("|autovalores| de B_J =", abs(eJ[0]))

DL_inv = np.linalg.inv(D+L)
B_GS = -np.dot(DL_inv, U)
eGS = np.linalg.eig(B_GS)
print("|autovalores| de B_GS =", abs(eGS[0]))
```
