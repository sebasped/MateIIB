# Librerías

```python
import numpy as np #para manejar arreglos
from matplotlib import pyplot as plt #para graficar

plt.rcParams["figure.figsize"] = (10,5) #tamaño de los gráficos: ancho x alto

from scipy.integrate import solve_ivp
# https://docs.scipy.org/doc/scipy/reference/generated/scipy.integrate.solve_ivp.html
```

# `solve_ivp` para resolver Ecuaciones Diferenciales.
## Utiliza un método como los que vimos (por defecto usa Runge-Kutta).

## Ejemplo: utilizando `solve_ivp` aproximar numéricamente la solución de:
### $$ x' = tcos(x(t))$$

```python
# definimos la función
def func(t,x):
    dx = t*np.cos(x)
    return dx

# condiciones iniciales e intervalo.
x0 = [1] #vector de condiciones iniciales
t0 = 0 #tiempo inicial
tf = 5 #tiempo final

# solve_ivp recibe: función, intervalo de tiempo y condicione iniciales
solucion = solve_ivp(func, [t0, tf], x0, dense_output = True)

# graficamos la solución
vals_t = np.linspace(t0,tf,1000)
vals_sol_x = solucion.sol(vals_t)
plt.plot(vals_t, vals_sol_x.T) # el .T es transponer. Es por como devuelve la solución solve_ivp
plt.grid() #marcas horizontales y verticales de refencia
```

## Mismo ejemplo pero resolviendo varias condiciones iniciales al mismo tiempo

```python
# definimos la función
def func(t,x):
    dx = t*np.cos(x)
    return dx

# condiciones iniciales e intervalo.
x0 = [1, 2, 3] #vector de condiciones iniciales
t0 = 0 #tiempo inicial
tf = 5 #tiempo final

# solve_ivp recibe: función, intervalo de tiempo y condicione iniciales
solucion = solve_ivp(func, [t0, tf], x0, dense_output = True)

# graficamos la solución
vals_t = np.linspace(t0,tf,1000)
vals_sol_x = solucion.sol(vals_t)
plt.plot(vals_t, vals_sol_x.T) # el .T es transponer. Es por como devuelve la solución solve_ivp
plt.grid() #marcas horizontales y verticales de refencia
```
