## Ejemplo 1:

### Se comercializan tres marcas A, B y C de un cierto producto. Desde el año 1990, todos los años,
- ### el 5% de los consumidores de A pasa a consumir B y el 15% de los consumidores de A pasa a consumir C;
- ### el 15% de los consumidores de B pasa a consumir C y el 5% de los consumidores de C pasa a consumir B,
- ### ningún consumidor de B o C vuelve a consumir A.

### 1. Hallar la matriz $P$ de transición del proceso.

### 2. ¿Cuántos estados de equilibrio hay y cuáles son?

### 3. ¿Todo estado inicial tiene estado límite?, ¿se puede decir cuáles son esos estados límite?, ¿depende el estado límite del estado inicial?

### 4.  Si al comenzar el año 1990 la mitad de los consumidores consumía la marca A y la otra mitad la marca B. Hallar el estado inicial $v(0)$. ¿Cuál es el estado límite para este $v(0)$?



```python
# 1

P = np.array([[0.8 ,0   ,0   ],
              [0.05,0.85,0.05],
              [0.15,0.15,0.95]])
```

```python
# 2

e = np.linalg.eig(P)
autovalores = e[0]
autovectores = e[1] # están en columnnas

print("Autovalores:",autovalores) # autovalores
print("Matriz de autovectores en columna:\n",autovectores) # autovects en columnas

# el estado de equilibrio es el autovect que corresponde al autoval=1
# pero hay que normalizarlo para que cumpla que sea estado de markov

# De paso observar que los otros autovects nunca pueden ser estados de Markov

# Chequeo que ese autovect es de equilibrio
estado_equilibrio = autovectores[:,0]
print("Estado de equilibrio (pero en realidad no es):", estado_equilibrio)

# Problema: no es un estado de Markov.
# Tienen que tener valores no negativos: esto está
# las entradas tienen que sumar 1: tengo que normalizar
estado_equilibrio_normalizado = estado_equilibrio / np.linalg.norm(estado_equilibrio, ord=1)
print("Estado de equilibrio ahora sí (después de normalizar):", estado_equilibrio_normalizado)

# Mini chequeo de que efectivamente es de equilibrio
print(np.matmul(P,estado_equilibrio_normalizado)) #debe dar igual que el estado
```

### 3: como 1 es autovalor simple: todo estado inicial tiene estado límite. Hacer mini cuenta de esto en papel.
### Y el estado límite es el único estado de equilibrio, que a su vez es el autovector (normalizado) correspondiente, es decir el hallado en el ítem anterior.

## Ejemplo 2:

### Se comercializan las mismas tres marcas, pero las reglas de transición ahora están dadas por la matriz
$$P =
\begin{pmatrix}
a & b & 0 \\
c & b & 0 \\
0 & 0 & 1
\end{pmatrix}$$

### 1. Sabiendo que $\det(P)=1/4$, hallar la matriz $P$ de transición del proceso e interpretarla.

### 2. ¿Cuántos estados de equilibrio hay y cuáles son?

### 3. ¿Todo estado inicial tiene estado límite?, ¿se puede decir cuáles son esos estados límite?, ¿depende el estado límite del estado inicial?

### 4. Hallar el estado límite para el estado inicial $v_0 = (1/2,1/3,1/6)$.


```python
# 1

P = np.array([[3/4,1/2,0],
              [1/4,1/2,0],
              [0  ,0  ,1]])
```

```python
# 2

e = np.linalg.eig(P)
autovalores = e[0]
autovectores = e[1] # están en columnnas

print("Autovalores:",autovalores) # autovalores
print("Matriz de autovectores en columna:\n",autovectores) # autovects en columnas

# los autovectores de equilibrio son
# v1 = e[1][:,0]
# v3 = e[1][:,2]
# print(v1)
# print(v3)

estado_eq1 = autovectores[:,0]
estado_eq1 = estado_eq1/sum(estado_eq1)

estado_eq2 = autovectores[:,2]
estado_eq2 = estado_eq1/sum(estado_eq2)

# OJO
# acá los autovects de autoval=1 son dos, y el otro autoval es de módulo menor a 1,
# con lo cual si bien todo estado inicial va a tener estado límite, van a haber muchos equilibrios:
# todas las combinaciones lineales de esos dos autovects (normalizados), con los coeficientes sumando 1 y ambos >= 0.
# por eso es que el estado límite va a depender del estado inicial
```

### 3: OJO acá los autovects de autoval=1 son dos, y el otro autoval es de módulo menor a 1, con lo cual si bien todo estado inicial va a tener estado límite, van a haber muchos equilibrios:
### todas las combinaciones lineales de esos dos autovects (normalizados), con los coeficientes sumando 1 y ambos >= 0.
### por eso es que el estado límite va a depender del estado inicial

```python
# 4

# Escribo el estado inicial como combinación lineal de los autovectores
v0 = np.array([1/2,1/3,1/6])
coefs_v0_base_autovectores = np.linalg.solve(autovectores,v0) #para hallar los coeficientes de la combinación de v0 con los autovects
# print(coefs_v0_base_autovectores)
# se puede experimentar cambiando el estado inicial para verificar que el estado límite cambia.

#calculamos vinf: lo correspondiente al autoval de módulo menor a 1 tiende a cero y por lo tanto no va.
v_infinito = coefs_v0_base_autovectores[0]*autovectores[:,0] + coefs_v0_base_autovectores[2]*autovectores[:,2]
print("Estado límite para estado inicial (1/2, 1/3, 1/6):",v_infinito)

# Chequeo que el estado sea de Markov: entradas no negativas y que sumen 1.
print("Chequeo que el estado límite está normalizado:",sum(v_infinito))


# También podemos verificar que el estado límite también es de equilibrio
print("Verificamos que el estado límite también es de equilibrio:",np.matmul(P,v_infinito))
```

## **Ejercicio:**
### Consideremos la matriz de Markov
$$P =
\begin{pmatrix}
0.8 & 0 & 0 \\
0.05 & 0.85 & 0.05 \\
0.15 & 0.15 & 0.95
\end{pmatrix}$$

### 1. Decidir si existe la matriz límite $P^{\infty}$.

### 2. De existir, calcular la matriz límite $P^{\infty}$.

### 3. ¿Que quedan en las columnas de $P^{\infty}$?


```python
# 1

P = np.array([[0.8 ,0   ,0   ],
              [0.05,0.85,0.05],
              [0.15,0.15,0.95]])

e = np.linalg.eig(P) # autovals y autovects

autovalores = e[0] # autovalores
autovectores = e[1] # matriz autovects en columnas

# Chequeo los autovalores para ver si existe la matriz límite
print(autovalores)
```

```python
# 2

# a partir de los autovalores defino la D infinito (ver cuenta en diapos Mercedes)
Dinf = np.diag([1,0,0]) #matriz con eso en la diagonal
# print(Dinf)

# necesito la inversa de la matriz de autovects
autovectores_inversa = np.linalg.inv(autovectores)

# Y ahora calculo la matriz de markov límite Pinf (ver cuenta en diapos Mercedes)
Pinf = np.linalg.multi_dot( [autovectores, Dinf, autovectores_inversa] )
print("Matriz límite:\n",Pinf)
```

```python
# 3
# En las columnas de Pinf queda el estado de equilibrio
print("Matriz límite:\n",Pinf)


# Chequeamos eso solamenta para verificarlo.
print("Autovalores:",autovalores) # autovalores
print("Matriz de autovectores en columna:\n",autovectores) # autovects en columnas

# estado equilibro: el asociado a autoval=1
estado_eq = autovectores[:,0]

# Chequeo que sea un estado de Markov.
# Tienen que tener valores no negativos: esto está
# las entradas tienen que sumar 1: tengo que normalizar
estado_eq = estado_eq/sum(estado_eq)
print("Estado de equilibrio:", estado_eq)
```
