## Ejemplo 1:

### Se comercializan tres marcas A, B y C de un cierto producto. Desde el año 1990, todos los años,
- ### el 5% de los consumidores de A pasa a consumir B y el 15% de los consumidores de A pasa a consumir C;
- ### el 15% de los consumidores de B pasa a consumir C y el 5% de los consumidores de C pasa a consumir B,
- ### ningún consumidor de B o C vuelve a consumir A.

### 1. Dibujar el diagrama de transición del proceso y hallar la matriz $P$ de transición del proceso.

### 2. Verificar que $v=(0,1/4,3/4)$ es un estado de equilibrio.

### 3.  Si al comenzar el año 1990 la mitad de los consumidores consumía la marca A y la otra mitad la marca B. Hallar el estado inicial $v(0)$, y calcular los estados $v(1), v(5), v(10), v(20), v(30), v(40), v(50)$.

### 4.  Analizando lo ocurrido en el ítem anterior, ¿podría darse cuenta de cuál es el estado límite de $v(0) = (1/2,1/2,0)$?

### 5.  Si al comenzar el año 1990 hay 10 mil consumidores, ¿podría darse cuenta del estado límite para la cantidad de individuos que consume cada marca?



```python
# 1
P = np.array([[0.8 ,0   ,0   ],
              [0.05,0.85,0.05],
              [0.15,0.15,0.95]])
```

```python
# 2
v = np.array([0, 1/4, 3/4])
w = np.dot(P,v)

# deben ser iguales
print(v)
print(w)
```

```python
# 3
v0 = np.array([1/2, 1/2, 0])
print("v0:", v0)

estados_de_interés = [1, 5, 10, 20, 30, 40, 50]

v = v0
for i in range(1,51,1):
  v_sig = np.dot(P,v)

  if i in estados_de_interés:
    print("v a tiempo ",i,"es",v_sig)

  v = v_sig

# ¿y si quiero el v50 de una?
P50 = np.linalg.matrix_power(P,50) #P multiplicada por sí misma 50 veces
v50 = np.dot(P50, v0)
print("v a tiempo 50 calculado de una",v50) #verificar con el anterior que da igual
```

### 4: el estado límite a ojo es el ...
### Comentar que luego lo averiguaremos de forma exacta teórica (por autovalores y autovectores de $P$) y que esto es una aproximación por experimentación.

```python
# 5: si el estado límite (0,1/4,3/4) son las proporciones límite de cada marca.
# entonces para averiguar las cantidades a partir de 10 mil es simplemente multiplicar

v_lim = np.array([0, 1/4, 3/4]) #calculado por ahora a ojo
pob_inicial = 10**4
pob_lim = pob_inicial*v_lim # número por vector
print("Población límite para cada marca:", pob_lim)
```
