```python
import numpy as np #para manejar arreglos
from matplotlib import pyplot as plt #para graficar

from numpy.fft import fft, fftfreq, ifft
# https://numpy.org/doc/stable/reference/routines.fft.html
# fft calcula la transformada
# fftfreq calcula las (posibles) frecuencias involucradas

from scipy.io import wavfile #leer archivos de audio wav

plt.rcParams["figure.figsize"] = (10,5) #tamaño de los gráficos: ancho x alto
```

```python
# cambiar por cualquiera que se quiera aproximar
def f_periodica(t):
  y = t**2
  return y
```

```python
freq_muestreo = 50
paso_temporal = 1/freq_muestreo
b = 5
vals_t = np.arange(0, b, paso_temporal) #inicio, fin, paso. No llega hasta el último
vals_f = f_periodica(vals_t)

print(len(vals_t))

plt.plot(vals_t, vals_f, label="f periódica")
```

```python
vals_f_fourier = fft(vals_f)

vals_f_recuperados = ifft(vals_f_fourier)

plt.plot(vals_t, vals_f_recuperados, label="f periódica")
```
